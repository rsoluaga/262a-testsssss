```python
import matplotlib.pyplot as plt

df ["Year Patron Registered"].value_counts().plot(
    kind="pie",
)

plt.title("Patron Registrations by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt

df ["Year Patron Registered"].value_counts().plot(
    kind="bar",
    color="purple"
)

plt.title("Patron Registrations by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt

df ["Circulation Active Month"].value_counts().plot(
    kind="bar",
    color="purple"
)

plt.title("Patron Registrations by Year")
plt.ylabel("# of Circulation")
plt.xlabel("month")
plt.show()
```

```python
month_order = [
    "Janurary", "Feburary", "March", "April", "June", "July", "August", "Septedmber", "October", "November", "December"
]

df ["Circulation Active Month"].value_counts().plot(
    kind="bar",
    color="purple",
    edgecolor="green",
    figsize=(11,8)
)

plt.title("Patron Registrations by Year")
plt.ylabel("# of Circulation")
plt.xlabel("month")
plt.show()
```

```python
df.isna().sum()
```

```python
df['Circulation Active Month Cleaned'] = df['Circulation Active Month'].fillna('Not Captured')

df.isna().sum()
```

```python
month_order = [
    "January", "February", "March", "April", "June", "July", "August", "September", "October", "November", "December", "Not Captured"
]

df ["Circulation Active Month Cleaned"].value_counts().reindex(month_order).plot(
    kind="bar",
    color="violet",
    edgecolor="orange",
    figsize=(11,8)
)

plt.title("Patron Registrations by Year")
plt.ylabel("# of Circulation")
plt.xlabel("month")
plt.show()
```
